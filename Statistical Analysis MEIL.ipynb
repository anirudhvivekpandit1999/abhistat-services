```python
import matplotlib.pyplot as plt
import seaborn as sns

# Function to plot the distribution with normalized weights
def plot_normalized_distribution(df_before, df_after):
    # Display available columns to the user
    print("Available columns for comparison:")
    available_columns = df_before.columns.intersection(df_after.columns).tolist()
    for idx, col in enumerate(available_columns, 1):
        print(f"{idx}. {col}")

    # Prompt user to select columns for the x and y axes
    x_input = input("Enter the column name or index for the X-axis: ").strip()
    y_input = input("Enter the column name or index for the Y-axis: ").strip()

    # Resolve the X-axis column
    if x_input.isdigit():
        x_index = int(x_input) - 1
        if 0 <= x_index < len(available_columns):
            x_column = available_columns[x_index]
        else:
            print("Invalid index for X-axis. Please try again.")
            return
    else:
        x_column = x_input
        if x_column not in available_columns:
            print(f"Column '{x_column}' not found. Please try again.")
            return

    # Resolve the Y-axis column
    if y_input.isdigit():
        y_index = int(y_input) - 1
        if 0 <= y_index < len(available_columns):
            y_column = available_columns[y_index]
        else:
            print("Invalid index for Y-axis. Please try again.")
            return
    else:
        y_column = y_input
        if y_column not in available_columns:
            print(f"Column '{y_column}' not found. Please try again.")
            return

    # Normalize the weights to compute the mean instead of the sum
    df_before["weighted_y"] = df_before[y_column] / df_before[y_column].sum()
    df_after["weighted_y"] = df_after[y_column] / df_after[y_column].sum()

    # Plot the histograms with normalized weights
    plt.figure(figsize=(10, 6))

    # Histogram for "before"
    sns.histplot(
        x=df_before[x_column],
        weights=df_before["weighted_y"],
        bins=30,
        label="Before",
        color="blue",
        alpha=0.6
    )

    # Histogram for "after"
    sns.histplot(
        x=df_after[x_column],
        weights=df_after["weighted_y"],
        bins=30,
        label="After",
        color="red",
        alpha=0.6
    )

    # Set labels and title
    plt.xlabel(x_column)
    plt.ylabel(f'Normalized {y_column}')
    plt.title(f'Normalized Distribution of {x_column} Weighted by {y_column} (Before and After)')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()

    plt.show()

# Call the function
plot_normalized_distribution(df_without_thermact_cleaned_selected, df_with_thermact_cleaned_selected)
```

## Confidence interval
1. What Does a Confidence Interval Represent?
A confidence interval provides a range of plausible values for the true difference in means (or another metric, such as proportions) between two groups (e.g., "before" and "after" the product was introduced).

If the true difference is 0, it means there is no difference between the two groups.
If the CI includes 0, we cannot rule out the possibility that there is no real difference; the observed difference may be due to random variation in the data.
If the CI does not include 0, it means the true difference is likely not 0, indicating a statistically significant impact.

```python
import numpy as np
from scipy.stats import t

# Function to calculate a 95% confidence interval
def calculate_confidence_interval(data, confidence=0.95):
    n = len(data)
    mean = np.mean(data)
    std_err = np.std(data, ddof=1) / np.sqrt(n)  # Standard error of the mean
    margin = std_err * t.ppf((1 + confidence) / 2., n - 1)  # Margin of error
    return mean, margin

# Function to analyze all columns for significant impact
def analyze_all_columns(df_before, df_after):
    # Find common numeric columns in both DataFrames
    common_columns = df_before.columns.intersection(df_after.columns)
    significant_impact = []
    no_significant_impact = []

    # Iterate through each column
    for column in common_columns:
        try:
            # Drop missing values
            data_before = df_before[column].dropna()
            data_after = df_after[column].dropna()

            # Skip non-numeric columns
            if not np.issubdtype(data_before.dtype, np.number):
                continue

            # Calculate confidence intervals
            mean_before, margin_before = calculate_confidence_interval(data_before)
            mean_after, margin_after = calculate_confidence_interval(data_after)

            # Determine confidence interval ranges
            lower_before, upper_before = mean_before - margin_before, mean_before + margin_before
            lower_after, upper_after = mean_after - margin_after, mean_after + margin_after

            # Check if confidence intervals overlap
            if upper_before < lower_after or upper_after < lower_before:
                significant_impact.append((column, mean_before, margin_before, mean_after, margin_after))
            else:
                no_significant_impact.append((column, mean_before, margin_before, mean_after, margin_after))
        except Exception as e:
            print(f"Error processing column '{column}': {e}")

    # Print results
    print("\nColumns with Significant Impact:")
    for column, mean_before, margin_before, mean_after, margin_after in significant_impact:
        print(f"{column}:")
        print(f"  Before: {mean_before:.3f} ± {margin_before:.3f}")
        print(f"  After: {mean_after:.3f} ± {margin_after:.3f}")
        print("  Significant impact detected.\n")

    print("\nColumns without Significant Impact:")
    for column, mean_before, margin_before, mean_after, margin_after in no_significant_impact:
        print(f"{column}:")
        print(f"  Before: {mean_before:.3f} ± {margin_before:.3f}")
        print(f"  After: {mean_after:.3f} ± {margin_after:.3f}")
        print("  No significant impact detected.\n")

# Call the function
analyze_all_columns(df_without_thermact_cleaned_selected, df_with_thermact_cleaned_selected)
```

```python
def calculate_percent_change(df_before, df_after):
    common_columns = df_before.columns.intersection(df_after.columns)
    results = []

    for column in common_columns:
        try:
            # Drop missing values and check for numeric columns
            data_before = df_before[column].dropna()
            data_after = df_after[column].dropna()
            if not np.issubdtype(data_before.dtype, np.number):
                continue

            # Calculate means
            mean_before = np.mean(data_before)
            mean_after = np.mean(data_after)

            # Calculate percent change
            percent_change = ((mean_after - mean_before) / mean_before) * 100
            results.append((column, mean_before, mean_after, percent_change))
        except Exception as e:
            print(f"Error processing column '{column}': {e}")

    # Print results
    print("\nImpact of the Product:")
    for column, mean_before, mean_after, percent_change in results:
        print(f"{column}:")
        print(f"  Mean Before: {mean_before:.3f}")
        print(f"  Mean After: {mean_after:.3f}")
        print(f"  Percent Change: {percent_change:.2f}%\n")

# Call the function
calculate_percent_change(df_without_thermact_cleaned_selected, df_with_thermact_cleaned_selected)
```

## Bootstrapping

```python
from sklearn.utils import resample
import numpy as np

def bootstrap_all_columns(df_before, df_after, n_bootstraps=10000):
    # Find common numeric columns
    common_columns = df_before.columns.intersection(df_after.columns)
    significant_impact = []
    no_significant_impact = []

    for column in common_columns:
        try:
            # Drop missing values and ensure numeric data
            data_before = df_before[column].dropna()
            data_after = df_after[column].dropna()

            if not np.issubdtype(data_before.dtype, np.number):
                continue

            # Generate bootstrap samples
            bootstrapped_differences = []
            for _ in range(n_bootstraps):
                sample_before = resample(data_before)
                sample_after = resample(data_after)
                bootstrapped_differences.append(np.mean(sample_after) - np.mean(sample_before))

            # Calculate confidence intervals
            lower_bound = np.percentile(bootstrapped_differences, 2.5)
            upper_bound = np.percentile(bootstrapped_differences, 97.5)
            mean_difference = np.mean(bootstrapped_differences)

            # Determine if the impact is significant
            if lower_bound > 0 or upper_bound < 0:
                significant_impact.append((column, mean_difference, lower_bound, upper_bound))
            else:
                no_significant_impact.append((column, mean_difference, lower_bound, upper_bound))
        except Exception as e:
            print(f"Error processing column '{column}': {e}")

    # Print results
    print("\nColumns with Significant Impact:")
    for column, mean_diff, lower, upper in significant_impact:
        print(f"{column}:")
        print(f"  Mean Difference: {mean_diff:.3f}")
        print(f"  95% Confidence Interval: [{lower:.3f}, {upper:.3f}]")
        print("  Statistically significant impact detected.\n")

    print("\nColumns without Significant Impact:")
    for column, mean_diff, lower, upper in no_significant_impact:
        print(f"{column}:")
        print(f"  Mean Difference: {mean_diff:.3f}")
        print(f"  95% Confidence Interval: [{lower:.3f}, {upper:.3f}]")
        print("  No statistically significant impact detected.\n")

# Usage Example:
bootstrap_all_columns(df_without_thermact_cleaned_selected, df_with_thermact_cleaned_selected)
```

```python
import scipy.stats as stats
import numpy as np

def analyze_column(column_name, df_before, df_after, confidence=0.98):
    # Extract data for the specified column
    before_data = df_before[column_name]
    after_data = df_after[column_name]
    
    # Calculate mean difference
    mean_before = np.mean(before_data)
    mean_after = np.mean(after_data)
    mean_difference = mean_after - mean_before
    
    # Calculate the standard error of the mean difference
    pooled_std = np.sqrt(np.var(before_data, ddof=1) / len(before_data) + 
                         np.var(after_data, ddof=1) / len(after_data))
    
    # Calculate the critical value (t-score)
    t_critical = stats.t.ppf((1 + confidence) / 2, df=len(before_data) + len(after_data) - 2)
    
    # Calculate confidence interval
    margin_of_error = t_critical * pooled_std
    lower_bound = mean_difference - margin_of_error
    upper_bound = mean_difference + margin_of_error
    
    # Analysis
    print(f"{column_name}:")
    print(f"  Mean Difference: {mean_difference:.3f}")
    print(f"  {int(confidence * 100)}% Confidence Interval: [{lower_bound:.3f}, {upper_bound:.3f}]")
    print("  Analysis:")
    if lower_bound <= 0 <= upper_bound:
        print("  - The confidence interval includes zero, indicating no statistically significant impact.")
    else:
        print("  - The confidence interval does not include zero, indicating a statistically significant impact.")
    
    if mean_difference > 0:
        print("  - There was an increase in this parameter.")
    elif mean_difference < 0:
        print("  - There was a decrease in this parameter.")
    print("\n")

# Example usage for specific columns
columns_to_analyze = [
    "Total SH Spray",  # Add specific column names
    "Total SH Flow",
    "Average APH",
    "Average O2 flow",
    "Column 211"
]

for col in columns_to_analyze:
    analyze_column(col, df_without_thermact_cleaned_selected, df_with_thermact_cleaned_selected)
```
